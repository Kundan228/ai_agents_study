```python
# input = "AaBb!abaCdc"
 
# Rules:
 
# Ignore special characters, digits, and spaces.
# Treat uppercase and lowercase letters as the same.
# Count the frequency of each alphabet.
# Sort by frequency (descending), then alphabetically.
# Append the total number of valid alphabets at the end as total<count>.
 
# Processing:
# a -> 4
# b -> 3
# c -> 2
# d -> 1
 
# Total alphabets = 10
 
# Expected output:
# a4b3c2d1total10
```

```python
input = "AaBb!abaCdc"

result = {}
for i in input.lower():
    if i.isalpha():
        # print(i)
        if i not in result:
            result[i] = 1
        else:
            result[i] = result[i] + 1

total = sum(list(result.values()))
re_temp = []
for i, v in result.items():
    re_temp.append(i)
    re_temp.append(str(v))
final_result = "".join(re_temp) + f"total{total}"
final_result
```

```python
# Check if two strings are anagrams
str1 = "listen"
str2 = "silent"

def ang(str1,str2):
    if "".join(sorted(str1)) == "".join(sorted(str2)):
        return True
    return False

ang(str1,str2)
```

```python
#Given a list of intervals, merge all overlapping intervals

intervals = [
    [1,3],
    [2,6],
    [8,10],
    [15,18]
]

re = []
for i in range(len(intervals)):
    if len(intervals) - 1 == i:
         break
    # print(intervals[i])
    if intervals[i][1:] > intervals[i+1][:1]:
        re.append([intervals[i][:1][0],intervals[i+1][1:][0]])
    else:
        re.append(intervals[i])
    # break

re
```

```python
# Group Employees by Department

# Given a list of employee records, group employees by department and calculate:
# Total employees
# Average salary

employees = [
    {"name":"John","dept":"IT","salary":50000},
    {"name":"Alice","dept":"HR","salary":40000},
    {"name":"Bob","dept":"IT","salary":70000},
    {"name":"Mike","dept":"HR","salary":60000},
    {"name":"David","dept":"IT","salary":90000}
]

import pandas as pd 
df = pd.DataFrame(employees)
df
```

```python
result = df.groupby("dept")["salary"].agg(
    count="count",
    average_salary="mean"
)

print(result)
```

```python
result = {}

for i in employees:
    dept = i["dept"]
    if result :
        if "IT" in result and "HR" in result:
            count = result[dept].get("count") + 1
            total_salary = result[dept].get("total_salary") + i['salary']
    else:
        count = 1
        total_salary = i['salary']
    result[dept] = {"count":count, "total_salary": total_salary}

result
```

```python
for i in result.keys():
    print(result[i]["total_salary"])
    result[i]["total_salary"] = result[i]["total_salary"]/ result[i]["count"]
```
